```python
def countdown():
    print("Countdown Started ...")
    for i in range(5, 0, -1):
        yield i
        
    print("Countdown Ended ...")

result=countdown()
print(next(result))
print(next(result))
print(next(result))
print(next(result))
print(next(result))
```

```python
logs = [
    "INFO: Application started \n",
    "INFO: User logged in \n",
    "WARNING: High memory usage\n",
    "ERROR: Database connection failed\n",
    "INFO: Application recovered\n",
]

def log_generation(logs):
    for log in logs :
        print("GENERATING LOG ...")
        yield log

    print("log generation completed...")

results=log_generation(logs)
for log in logs:
    print(next(results))
```

```python
logs = [
    "INFO: Application started\n",
    "INFO: User logged in\n",
    "WARNING: High memory usage\n",
    "ERROR: Database connection failed\n",
    "INFO: Application recovered\n",
    "ERROR: Payment processing failed\n",
    "WARNING: Disk space low\n",
    "ERROR: Authentication service unavailable\n"
]
def filter_error_logs(logs):
    for log in logs:
        if "ERROR" in log:
            print("Generating log...")
            yield log

results=filter_error_logs(logs)
try:
    for log in results:
        print(log)
except StopIteration:
    print("NO MORE LOGS")
```

```python
transactions = [
    {"id": 101, "amount": 500},
    {"id": 102, "amount": 2500},
    {"id": 103, "amount": 750},
    {"id": 104, "amount": 5000},
    {"id": 105, "amount": 1200},
    {"id": 106, "amount": 300},
    {"id": 107, "amount": 3500},
]
def process_transactions(transactions):
    for t in transactions:
        print("Checking transaction...")
        if t["amount"]>1000:
            yield t

    print("Transaction processing completed")


results = process_transactions(transactions)

for transaction in results:
    print(
        f"Transaction ID: {transaction['id']}, "
        f"Amount: {transaction['amount']}"
    )
```

```python
prices = [500, 1200, 2500, 800, 4000, 1500]
def  Discounted_prices(prices):
    for price in prices:
        print("calculating price...")
        if price>1000:
            yield price-float(price*0.10)
    print("NO more products...")

results= Discounted_prices(prices)

for price in results:
    print(price)
```
