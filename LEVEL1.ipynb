```python
#Lesson 1.3 - Conditional logic with multiple conditions , example - login eligibility system.
u=input("Enter username:")
pw= input("Enter Password:")
print("Access Granted " if u=="admin" and pw=="python123" else "Access denied")
```

```python
#Online discount system
is_premium = input("Are you premium? enter YES or NO:")
amt=int(input("Enter amount:"))
print("Discount applied " if is_premium=="YES" or amt >=5000 else "DISCOUNT NOT APPLIED")
```

```python
#Lesson 1.4 - The not operator , Example- Real-world scenario: Library book availability
# Library book availability
is_borrowed = False
print("Book Available" if not is_borrowed else "Book Not Available")
```

```python
#Level 1 — Final Mini Project
#Project: Smart Shopping Eligibility System
#Imagine you're building a shopping application.
#The application must determine the customer's discount category.
##Business rules
#Premium customer AND purchase ≥ ₹5,000 → "VIP Discount"
#Premium customer OR purchase ≥ ₹3,000 → "Special Discount"
#Purchase ≥ ₹1,000 → "Regular Discount"
#Otherwise → "No Discount"

is_premium = input("Are you premium? enter YES or NO:")
amt=int(input("Enter amount:"))
print("VIP Discount" if is_premium.upper()=="YES" and amt >= 5000 else "Special Discount" if is_premium.upper()=="YES" or amt >= 5000 else "No Discount" )
```
