```python
#Level 2 - Comprehensions
#Task 1: Employee Data Processing System
#Imagine you're working on a backend project.
#Your application receives employee records and needs to process them into different data structures.

#INPUT
employees = [
    {"name": "Alice", "age": 28, "salary": 75000, "department": "IT",
     "skills": ["Python", "SQL", "Django"]},

    {"name": "Bob", "age": 35, "salary": 95000, "department": "HR",
     "skills": ["Recruitment", "Communication"]},

    {"name": "Charlie", "age": 24, "salary": 55000, "department": "IT",
     "skills": ["Python", "Java"]},

    {"name": "David", "age": 40, "salary": 120000, "department": "Finance",
     "skills": ["Excel", "Accounting"]},

    {"name": "Eva", "age": 29, "salary": 85000, "department": "IT",
     "skills": ["Python", "SQL", "Flask"]},

    {"name": "Frank", "age": 32, "salary": 65000, "department": "HR",
     "skills": ["Communication", "Management"]},

    {"name": "Grace", "age": 26, "salary": 45000, "department": "Finance",
     "skills": ["Excel", "Accounting"]}
]
```

```python
#Task 1 - List comprehension
#Create a list containing the names of employees whose salary is greater than 70,000.
lst=[employee["name"] for employee in employees if employee["salary"]>70000] 
print(lst)
```

```python
#Task 2 - Conditional comprehension
#Requirement: Create a list of employee names with their salary category.
categories = [
    (
        employee["name"],
        "High" if employee["salary"]>=90000 else "Medium" if employee["salary"]>=60000 else "Low"
    )
    for employee in employees
]
print(categories)
```

```python
#Task 3 - Set Comprehension
#Create a set containing all unique department names
dept = {employee["department"] for employee in employees }
print("DEPARTMENTS :", dept)
```

```python
#Task 4 - Dictionary Comprehension
#calculating number of alphabets in employee name
dct={ employee["name"]:len(employee["name"]) for employee in employees }
print(dct)
```

```python
#Task 5 - Dictionary Comprehension - Employee Salary Increment 10%
salary_increment = {employee["name"]:int(employee["salary"]*1.10) for employee in employees }
print(salary_increment)
```

```python
#Task 5 - Nested Comprehensions
#Challenge 1 - Flatten employee skills
skill=[skill for employee in employees for skill in employee["skills"]]

#Challenge 2 — Filter skills -Extract only skills belonging to IT department employees.
it_skills = [skill for employee in employees for skill in employee["skills"] if employee["department"]=="IT" ]
print("IT SKILLS :", it_skills)

#Challenge 3 - Advanced nested comprehension - Create a list of tuples containing employee names and each of their skills.
emp_n_s = [(employee["name"], i) for employee in employees for i in employee["skills"]]
print("\n" , emp_n_s)

#Challenge 4 - Unique skills - Create a set of all unique skills across the company.
unique_skills = {s for s in skill}
print("\n Skills", unique_skills)
```
