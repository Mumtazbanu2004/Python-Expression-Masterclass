```python
# find cube of a number
cube = lambda x: x**3
num = int(input("Enter a Number:"))
print(f"Cube of the number {num}: {cube(num)}")
```

```python
#Average of three numbers
avg= lambda a,b,c : (a+b+c)/3
a, b, c = map(int, input("Enter 3 numbers separated by spaces: ").split())
print(f"Average of the three numbers {a},{b},{c}: {avg(a,b,c)}")
```

```python
employees = [
    {"name": "John", "age": 25, "department": "IT", "salary": 45000, "performance": 85},
    {"name": "Sarah", "age": 30, "department": "HR", "salary": 55000, "performance": 92},
    {"name": "David", "age": 22, "department": "IT", "salary": 30000, "performance": 65},
    {"name": "Mike", "age": 35, "department": "Finance", "salary": 75000, "performance": 95},
    {"name": "Emma", "age": 28, "department": "IT", "salary": 60000, "performance": 78},
    {"name": "Robert", "age": 40, "department": "Finance", "salary": 90000, "performance": 88},
    {"name": "Lisa", "age": 26, "department": "HR", "salary": 40000, "performance": 70}
]
```

```python
#Challenge 1 - Filter employees
is_IT_employees= list (filter(lambda emp:emp["department"]=="IT" and emp["performance"]>=80,employees))
print("IT Employees with performance equal or above 80\n",is_IT_employees)
```

```python
#Challenge 2 - Salary increment
#Use map() and lambda to increase every employee's salary by 10%.
updated_employees = list(map(lambda emp: {**emp,"salary": emp["salary"] * 1.10},employees))
print(updated_employees)
```

```python
#Challenge 3 - Performance ranking
#Use sorted() and lambda to sort employees according to performance
performance_ranking = sorted(
    employees,
    key=lambda emp: emp["performance"],
    reverse=True
)

print(performance_ranking)
```

```python
high_earners = list(
    filter(
        lambda emp: emp["salary"] > 50000,
        employees
    )
)

print(high_earners)
```

```python
employee_report = list(
    map(
        lambda emp: (emp["name"], emp["performance"]),
        employees
    )
)

print(employee_report)
```
